```python
from pathlib import Path
import sys
import importlib

# Locate this scripts/ folder even if the kernel cwd is elsewhere.
_MARKER = "figure_paths.py"
_candidates = [
    Path.cwd(),
    Path.cwd() / "scripts",
    Path.cwd() / "mn5sync_pdfba/sample_projects_intracellular/fba/cancer_core_metabolism/scripts",
    Path.cwd() / "../../mn5sync_pdfba/sample_projects_intracellular/fba/cancer_core_metabolism/scripts",
]
for parent in Path.cwd().resolve().parents:
    _candidates.append(
        parent / "mn5sync_pdfba/sample_projects_intracellular/fba/cancer_core_metabolism/scripts"
    )

SCRIPTS = next((c.resolve() for c in _candidates if (c / _MARKER).is_file()), None)
if SCRIPTS is None:
    raise FileNotFoundError(
        f"Could not find scripts/{_MARKER} from cwd={Path.cwd()}"
    )

# Prefer this scripts/ over any other figure_generation on sys.path
sys.path = [p for p in sys.path if Path(p).resolve() != SCRIPTS]
sys.path.insert(0, str(SCRIPTS))

import figure_paths
importlib.reload(figure_paths)

SCRIPTS = figure_paths.find_scripts_dir()
figure_paths.ensure_paths(SCRIPTS)

import figure_generation
importlib.reload(figure_generation)

print("scripts:", SCRIPTS)
print("figure_generation:", figure_generation.__file__)
print("default output:", figure_paths.default_output_dir())
```

## Main figures

1. Growth-rate map at `TIME_POINT`
2. Net export rates grid (uptake / secretion profiles only — no concentration overlay)


```python
import importlib
import figure_generation
importlib.reload(figure_generation)

# 1) Growth-rate spatial map
rc = figure_generation.main([
    "--output-dir", str(OUTPUT_DIR),
    "--fig-dir", str(FIG_DIR),
    "--time-point", str(TIME_POINT),
    "--panels",
    "--only", "growth_rate",
    "--format", "svg",
])
assert rc == 0, f"growth-rate panel failed with code {rc}"
growth_svg = FIG_DIR / f"growth_rate_{TIME_POINT}h.svg"
print("growth rate:", "ok" if growth_svg.is_file() else "not found", "→", growth_svg)

# 2) Main composite: net export / uptake profiles vs distance
rc = figure_generation.main([
    "--output-dir", str(OUTPUT_DIR),
    "--fig-dir", str(FIG_DIR),
    "--time-point", str(TIME_POINT),
    "--net-export-rates",
    "--format", "svg",
])
assert rc == 0, f"net-export grid failed with code {rc}"
export_svg = FIG_DIR / f"net_export_rates_grid_{TIME_POINT}h.svg"
print("net export grid:", "ok" if export_svg.is_file() else "not found", "→", export_svg)
```
